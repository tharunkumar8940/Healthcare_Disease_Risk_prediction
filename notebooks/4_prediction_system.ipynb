```python
import pandas as pd
import numpy as np
import joblib
```

```python
model = joblib.load(
    "../models/healthcare_disease_risk_model.pkl"
)

print("Model loaded successfully!")
```

```python
scaler = joblib.load(
    "../models/healthcare_scaler.pkl"
)

print("Scaler loaded successfully!")
```

```python
import pandas as pd
import joblib
import os

train_data = pd.read_csv(
    "../data/processed/train_data.csv"
)

feature_names = train_data.drop(
    "Disease_Risk",
    axis=1
).columns.tolist()

os.makedirs("../models", exist_ok=True)

joblib.dump(
    feature_names,
    "../models/feature_names.pkl"
)

print("feature_names.pkl created successfully!")
```

```python
feature_names = joblib.load(
    "../models/feature_names.pkl"
)

print("Feature names loaded successfully!")
```

```python
patient = {
    "Age": 45,
    "BMI": 27.5,
    "BloodPressure": 130,
    "Cholesterol": 210,
    "Glucose": 115,
    "HeartRate": 78,
    "Gender": "Male",
    "Smoking": "No",
    "PhysicalActivity": "Medium",
    "FamilyHistory": "Yes"
}

print(patient)
```

```python
patient_df = pd.DataFrame([patient])

patient_df
```

```python
patient_encoded = pd.get_dummies(
    patient_df,
    columns=[
        "Gender",
        "Smoking",
        "PhysicalActivity",
        "FamilyHistory"
    ],
    drop_first=True
)

patient_encoded
```

```python
patient_encoded = patient_encoded.reindex(
    columns=feature_names,
    fill_value=0
)

print("Patient features:", patient_encoded.shape)
print(patient_encoded)
```

```python
patient_scaled = scaler.transform(
    patient_encoded
)

print("Patient data scaled successfully!")
```

```python
prediction = model.predict(
    patient_scaled
)

print("Prediction:", prediction[0])
```

```python
if prediction[0] == 1:
    risk = "High Risk"
else:
    risk = "Low Risk"

print("Disease Risk:", risk)
```

```python
probability = model.predict_proba(
    patient_scaled
)[0][1]

print(
    "Risk Probability:",
    round(probability * 100, 2),
    "%"
)
```

```python
def predict_disease_risk(patient):

    patient_df = pd.DataFrame([patient])

    patient_encoded = pd.get_dummies(
        patient_df,
        columns=[
            "Gender",
            "Smoking",
            "PhysicalActivity",
            "FamilyHistory"
        ],
        drop_first=True
    )

    patient_encoded = patient_encoded.reindex(
        columns=feature_names,
        fill_value=0
    )

    patient_scaled = scaler.transform(
        patient_encoded
    )

    prediction = model.predict(
        patient_scaled
    )[0]

    probability = model.predict_proba(
        patient_scaled
    )[0][1]

    if prediction == 1:
        risk = "High Risk"
    else:
        risk = "Low Risk"

    return risk, probability
```

```python
risk, probability = predict_disease_risk(patient)

print("================================")
print("     DISEASE RISK RESULT")
print("================================")
print("Risk:", risk)
print("Probability:", round(probability * 100, 2), "%")
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output
```

```python
age = widgets.IntText(
    description="Age:",
    value=45
)

gender = widgets.Dropdown(
    description="Gender:",
    options=["Male", "Female"],
    value="Male"
)

bmi = widgets.FloatText(
    description="BMI:",
    value=27.5
)

blood_pressure = widgets.IntText(
    description="BP:",
    value=130
)

cholesterol = widgets.IntText(
    description="Cholesterol:",
    value=210
)

glucose = widgets.IntText(
    description="Glucose:",
    value=115
)

heart_rate = widgets.IntText(
    description="Heart Rate:",
    value=78
)

smoking = widgets.Dropdown(
    description="Smoking:",
    options=["Yes", "No"],
    value="No"
)

physical_activity = widgets.Dropdown(
    description="Activity:",
    options=["Low", "Medium", "High"],
    value="Medium"
)

family_history = widgets.Dropdown(
    description="Family History:",
    options=["Yes", "No"],
    value="Yes"
)
```

```python
predict_button = widgets.Button(
    description="Predict Disease Risk",
    button_style="primary"
)

output = widgets.Output()
```

```python
def on_predict_clicked(button):

    with output:
        clear_output()

        patient = {
            "Age": age.value,
            "BMI": bmi.value,
            "BloodPressure": blood_pressure.value,
            "Cholesterol": cholesterol.value,
            "Glucose": glucose.value,
            "HeartRate": heart_rate.value,
            "Gender": gender.value,
            "Smoking": smoking.value,
            "PhysicalActivity": physical_activity.value,
            "FamilyHistory": family_history.value
        }

        risk, probability = predict_disease_risk(patient)

        print("=" * 40)
        print("       DISEASE RISK RESULT")
        print("=" * 40)

        print("Risk:", risk)
        print(
            "Probability:",
            round(probability * 100, 2),
            "%"
        )

        print("=" * 40)
```

```python
predict_button.on_click(on_predict_clicked)
```

```python
display(
    age,
    gender,
    bmi,
    blood_pressure,
    cholesterol,
    glucose,
    heart_rate,
    smoking,
    physical_activity,
    family_history,
    predict_button,
    output
)
```
